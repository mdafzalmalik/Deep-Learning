```python
import numpy as np
from sklearn.datasets import make_classification
```

```python
X, y = make_classification(
    n_samples = 100,
    n_features = 2,
    n_informative = 1,
    n_redundant = 0,
    n_classes = 2,
    random_state = 42,
    n_clusters_per_class = 1,
    hypercube = False,
    class_sep = 2
)
```

```python
y
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.scatterplot(x = X[:, 0], y = X[:, 1], hue = y)
```

```python
def perceptron(X, y):
    weights = np.ones(X.shape[1])
    bias = 1.
    lr = 0.1

    for i in range(1000):
        j = np.random.randint(0, 100)
        weighted_sum = np.dot(X[j], weights)+bias
        y_hat = np.where(weighted_sum >= 0, 1, 0)
        weights = weights + lr * (y[j]-y_hat) * X[j]
        bias = bias + lr * (y[j]-y_hat)
    
    return weights, bias
```

```python
weights, bias = perceptron(X, y)
```

```python
weights
```

```python
bias
```

```python
m = -(weights[0]/weights[1])
b = -(bias/weights[1])
```

```python
m, b
```

```python
x_input = np.linspace(-3, 3, 100)
```

```python
y_input = m*x_input+b
```

```python
sns.lineplot(x = x_input, y = y_input)
sns.scatterplot(x = X[:, 0], y = X[:, 1], hue = y)
plt.show()
```
